# inference on video

```python
cap = cv2.VideoCapture('F:/video3.mp4')  # for Videog
```

```python
model = YOLO(r"D:\testing\finalShieldVision\runs\detect\train\weights\best.pt")
```

```python
classNames = ['gun']
```

```python


output_filename = 'output_video3.avi'
frame_width, frame_height = int(cap.get(3)), int(cap.get(4))
fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter(output_filename, fourcc, 20.0, (frame_width, frame_height))

while True:
    # Capture frame-by-frame
    success, img = cap.read()
    if not success:
        break  # Exit the loop if no frames are left

    # Perform detection using the YOLO model
    results = model(img, stream=True)
    for r in results:
        boxes = r.boxes
        for box in boxes:
            # Extract bounding box coordinates
            x1, y1, x2, y2 = box.xyxy[0]  # Get coordinates of the box
            x1, y1, x2, y2 = int(x1), int(y1), int(x2), int(y2)
            w, h = x2 - x1, y2 - y1

            # Draw bounding box on the image
            cvzone.cornerRect(img, (x1, y1, w, h), colorR=(255, 0, 0))

            # Display confidence score
            conf = math.ceil((box.conf[0] * 100)) / 100
            cls = int(box.cls[0])  # Get the class index
            cvzone.putTextRect(img, f'{classNames[cls]} {conf}', (max(0, x1), max(35, y1)), scale=1, thickness=1)

    # Write the processed frame to the output video
    out.write(img)

    # Display the frame in a windowgggggg
    cv2.imshow('Detection Output', img)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break  # Exit the loop when 'q' key is pressed

# Release resources
cap.release()
out.release()
cv2.destroyAllWindows()
```
