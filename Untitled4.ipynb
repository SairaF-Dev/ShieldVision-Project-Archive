```python
from pydrive.auth import GoogleAuth
from pydrive.drive import GoogleDrive

gauth = GoogleAuth()
gauth.LoadClientConfigFile(client_secret_path)

cred_file = r"D:\SheildVisionAlert\ShieldVision_assets\mycreds.txt"

# If credentials file doesn't exist OR token is expired, login again
try:
    gauth.LoadCredentialsFile(cred_file)
    if gauth.credentials is None:
        print("No credentials found. Logging in.")
        gauth.LocalWebserverAuth()
    elif gauth.access_token_expired:
        print("Token expired. Refreshing...")
        gauth.Refresh()
    else:
        gauth.Authorize()
except:
    print("Token invalid or revoked. Forcing re-authentication.")
    gauth.LocalWebserverAuth()

# Save the new credentials
gauth.SaveCredentialsFile(cred_file)

# Use drive
drive = GoogleDrive(gauth)
```

```python
# ---------------- Load YOLO Model ----------------
model = YOLO(r"D:\SheildVisionAlert\train2\weights\best.pt")
classNames = ['gun']
```

```python
# ---------------- Capture Video ----------------
cap = cv2.VideoCapture(r"D:\SheildVisionAlert\gunVideo.mp4")
```

```python
# ---------------- Location Functions ----------------
def get_location():
    g = geocoder.ip('me')
    return g.latlng if g.latlng else (None, None)

def get_address(lat, lng):
    if lat and lng:
        return f"https://www.google.com/maps?q={lat},{lng}"
    return "Location not available"
```

```python
# ---------------- Upload Frame to Google Drive ----------------
def upload_to_drive(image_path):
    try:
        file = drive.CreateFile({'title': 'detected_frame.jpg'})
        file.SetContentFile(image_path)
        file.Upload()
        file.InsertPermission({'type': 'anyone', 'value': 'anyone', 'role': 'reader'})
        return file['alternateLink']
    except Exception as e:
        print(f"❌ Google Drive upload failed: {e}")
        return None
```

```python
# ---------------- WhatsApp Alert ----------------
def send_whatsapp_alert(phone_number, image_url, location_url, time, date):
    url = f"https://graph.facebook.com/v21.0/{whatsapp_phone_id}/messages"
    print(f"📡 Sending request to: {url}")
    headers = {
        'Authorization': whatsapp_token,
        'Content-Type': 'application/json'
    }

    image_payload = {
        "messaging_product": "whatsapp",
        "to": phone_number,
        "type": "image",
        "image": {"link": image_url}
    }
    response_img = requests.post(url, json=image_payload, headers=headers)
    print("📸 WhatsApp Image Response:", response_img.status_code, response_img.text)

    text_payload = {
        "messaging_product": "whatsapp",
        "to": phone_number,
        "type": "text",
        "text": {
            "body": f"⚠ Security Alert: Gun Detected\n\n📍 Location: {location_url}\n⏰ Time: {time}\n📅 Date: {date}\n📷 Evidence: {image_url}\n\n*Immediate action required!*"
        }
    }
    response_txt = requests.post(url, json=text_payload, headers=headers)
    print("💬 WhatsApp Text Response:", response_txt.status_code, response_txt.text)
```

```python
# ---------------- Email Alert ----------------
def send_email_alert(from_email, password, to_email, subject, body, attachment_path):
    try:
        msg = EmailMessage()
        msg['Subject'] = subject
        msg['From'] = from_email
        msg['To'] = to_email
        msg.set_content(body)
        with open(attachment_path, 'rb') as f:
            img_data = f.read()
            msg.add_attachment(img_data, maintype='image', subtype='jpeg', filename='detected_frame.jpg')
        with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp:
            smtp.login(from_email, password)
            smtp.send_message(msg)
        print("✅ Email alert sent successfully.")
    except Exception as e:
        print(f"❌ Failed to send email: {e}")
```

```python
# ---------------- SMS Alert (Twilio) ----------------
def send_sms_alert(location_url, time_str, date_str, image_url, to_number):
    print("🛠️ Entered send_sms_alert()")  # Debug line 1

    message_body = (
        f"🔫 Gun Detected!\n"
        f"📷 Image: {image_url}"
    )

    print("📨 SMS Message Body Prepared")  # Debug line 2

    try:
        client = Client(TWILIO_SID, TWILIO_AUTH)
        print("✅ Twilio client initialized")  # Debug line 3

        message = client.messages.create(
            body=message_body,
            from_=TWILIO_NUM,
            to=to_number
        )
        print("✅ SMS sent! SID:", message.sid)  # Debug line 4
    except Exception as e:
        print(f"❌ SMS failed: {e}")  # Debug line 5
```

```python
# ---------------- Delete Old Files (>2 days) ----------------
def delete_old_files_by_days(folder_path, max_age_days=2):
    current_time = time.time()
    max_age_seconds = max_age_days * 86400
    for file in glob.glob(os.path.join(folder_path, "*.jpg")):
        if os.path.getctime(file) < (current_time - max_age_seconds):
            try:
                os.remove(file)
                print(f"🗑 Deleted: {file}")
            except Exception as e:
                print(f"❌ Deletion error: {e}")
```

```python
# ---------------- Main Detection Loop ----------------
last_alert_time = 0
alert_cooldown = 60  # seconds

while True:
    success, img = cap.read()
    if not success:
        break
    results = model(img, stream=True)
    for r in results:
        boxes = r.boxes
        for box in boxes:
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            w, h = x2 - x1, y2 - y1
            cvzone.cornerRect(img, (x1, y1, w, h), colorR=(255, 0, 0))
            conf = math.ceil((box.conf[0] * 100)) / 100
            cls = int(box.cls[0])
            label = classNames[cls]
            cvzone.putTextRect(img, f'{label} {conf}', (max(0, x1), max(35, y1)), scale=1, thickness=1)

            if label == 'gun' and (time.time() - last_alert_time > alert_cooldown):
                now = datetime.now()
                timestamp = now.strftime("%Y%m%d_%H%M%S")
                image_path = os.path.join(assets_folder, f"detected_{timestamp}.jpg")
                cv2.imwrite(image_path, img)

                delete_old_files_by_days(assets_folder, max_age_days=2)

                drive_link = upload_to_drive(image_path)
                if drive_link:
                    match = re.search(r'/d/([a-zA-Z0-9_-]+)', drive_link)
                    file_id = match.group(1) if match else None
                    direct_image_url = f"https://drive.google.com/uc?export=view&id={file_id}" if file_id else drive_link

                    lat, lng = get_location()
                    location_url = get_address(lat, lng)
                    time_str = now.strftime("%H:%M:%S")
                    date_str = now.strftime("%Y-%m-%d")

                    # WhatsApp Alerts
                    recipients = whatsapp_to.split(',')
                    for number in recipients:
                        send_whatsapp_alert(number.strip(), direct_image_url, location_url, time_str, date_str)

                    # SMS Alerts
                    sms_recipients = SMS_TO.split(',')
                    for number in sms_recipients:
                        send_sms_alert(location_url, time_str, date_str, direct_image_url, number.strip())

                    # Email Alert
                    subject = "🚨 Gun Detected Alert"
                    body = f"""🚨 A gun has been detected!

📍 Location: {location_url}
🕒 Time: {time_str}
📅 Date: {date_str}
📷 Image: {drive_link}

Immediate action is recommended.
"""
                    send_email_alert(from_email, password, to_email, subject, body, image_path)
                    last_alert_time = time.time()

    cv2.imshow('Detection Output', img)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
